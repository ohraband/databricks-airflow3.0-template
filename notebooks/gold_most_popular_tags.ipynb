```python
stg_posts_df = spark.table("data_platform_tutorial.default.stg_posts")
```

```python
from pyspark.sql import DataFrame
import pyspark.sql.functions as F

def posts_top_tags(stg_posts_df: DataFrame) -> DataFrame:
    return(
        stg_posts_df
        .withColumn("tag_exploded", F.explode("TagsArray"))
        .groupBy("tag_exploded")
        .agg(F.approx_count_distinct("PostId").alias("tags_count"))
        .orderBy(F.desc("tags_count"))
    )

marts_top_tags_df = posts_top_tags(stg_posts_df)
marts_top_tags_df.write.mode("overwrite").saveAsTable("data_platform_tutorial.default.marts_top_tags")
```
